```python
import plotly.graph_objects as go

from awp2.config import BAND_PREFIX, SEED
from awp2.data import load_split, wavelengths
from awp2.preprocessing import PreprocessingConfig, build_preprocessor

split = load_split()
sample = split.X_train.sample(n=5, random_state=SEED)
plain = build_preprocessor(PreprocessingConfig(use_meta=False)).fit(split.X_train)
smoothed = build_preprocessor(PreprocessingConfig(use_meta=False,
                                                  use_savgol=True,
                                                  savgol_window_length=5,
                                                  savgol_polyorder=2)).fit(split.X_train)

before = plain.transform(sample)
after = smoothed.transform(sample)
bands = [c for c in before.columns if c.startswith(BAND_PREFIX)]
x = wavelengths(before[bands])

colors = ("#636EFA", "#EF553B", "#00CC96", "#AB63FA", "#FFA15A")
fig = go.Figure()
for position, index in enumerate(sample.index):
    color = colors[position % len(colors)]
    hover = "Wellenlänge=%{x} nm<br>Reflektanz=%{y}<extra></extra>"
    fig.add_trace(go.Scatter(x=x, y=before.loc[index, bands].values,
                             mode="lines", name=f"{index} (roh)",
                             line={"color": color}, hovertemplate=hover))
    fig.add_trace(go.Scatter(x=x, y=after.loc[index, bands].values,
                             mode="lines", name=f"{index} (geglättet)",
                             line={"color": color, "dash": "dash"},
                             hovertemplate=hover))

fig.update_layout(title="Savitzky-Golay: 5 Spektren vorher/nachher",
                  xaxis={"title": "Wellenlänge [nm]", "type": "linear"},
                  yaxis_title="Reflektanz [%]")
fig.show()
```
