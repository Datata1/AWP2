# Kombinierte Labels: Monatskodierung für die RBF-SVM

**Hypothese:** Eine kategorische oder zyklische Darstellung von `Month` beschreibt den saisonalen Kontext besser als die bestehende numerische Skalierung und verbessert dadurch die kombinierte Balanced Accuracy.

**Eingaben:** Skalierte Spektren, One-Hot-kodierte AEZ und genau eine von drei Monatsdarstellungen (`numeric`, `one_hot`, `cyclical`).

**Akzeptanzkriterium:** Ausschließlich die höchste `bacc_combined` auf den fünf Trainingsfolds bestimmt die Monatsdarstellung und die SVM-Hyperparameter. Nur dieser Kandidat wird einmal auf dem gemeinsamen Validierungssplit bewertet.

```python
%load_ext autoreload
%autoreload 2

import pandas as pd
from IPython.display import display

from awp2.config import (
    MONTH_ENCODING_CYCLICAL,
    MONTH_ENCODING_NUMERIC,
    MONTH_ENCODING_ONE_HOT,
)
from awp2.experiment import RunConfig, TuneConfig, run, tune
from awp2.models import make_combined_svm
from awp2.preprocessing import PreprocessingConfig

month_encodings = (
    MONTH_ENCODING_NUMERIC,
    MONTH_ENCODING_ONE_HOT,
    MONTH_ENCODING_CYCLICAL,
)
```

## 1. Hyperparameter und Monatsdarstellung suchen

Jede Monatsdarstellung erhält dieselbe kleine Suche für `C` und `gamma`. Die Vergleiche nutzen ausschließlich die festen Trainingsfolds; weder der gemeinsame Validierungssplit noch dessen Referenzscore gehen in die Auswahl ein.

```python
tuned_by_encoding = {}

for month_encoding in month_encodings:
    preprocessing = PreprocessingConfig(
        scale=True,
        use_meta=True,
        month_encoding=month_encoding,
    )
    tuned_by_encoding[month_encoding] = tune(
        make_combined_svm(),
        TuneConfig(
            name=f"combined_svm_month_{month_encoding}_search",
            approach="combined",
            description=(
                "RBF-SVM auf kombinierten Crop/Stage-Labels mit "
                f"{month_encoding} Month-Kodierung; C und gamma per CV vergleichen."
            ),
            preprocessing=preprocessing,
            param_grid={
                "estimator__C": [30.0, 100.0, 300.0],
                "estimator__gamma": [0.001, 0.003, 0.01],
            },
        ),
    )

cv_summary = pd.DataFrame(
    [
        {
            "Month-Kodierung": month_encoding,
            "CV BAcc kombiniert": tuned.best_score,
            "Beste Parameter": tuned.best_params,
        }
        for month_encoding, tuned in tuned_by_encoding.items()
    ]
).sort_values("CV BAcc kombiniert", ascending=False)
display(cv_summary)

best_month_encoding = cv_summary.iloc[0]["Month-Kodierung"]
best_tuned = tuned_by_encoding[best_month_encoding]
```

## 2. Einmal validieren

Nur die Monatsdarstellung mit der höchsten CV-BAcc wird einmal auf dem gemeinsamen Validierungssplit bewertet. Die beiden übrigen Kodierungen erhalten keine Holdout-Bewertung.

```python
best_preprocessing = PreprocessingConfig(
    scale=True,
    use_meta=True,
    month_encoding=best_month_encoding,
)

result = run(
    best_tuned.best_model,
    RunConfig(
        name="combined_svm_month_encoding",
        approach="combined",
        description=(
            "Beste Month-Kodierung aus drei CV-Suchen, einmal auf dem gemeinsamen "
            "Validierungssplit bewertet."
        ),
        preprocessing=best_preprocessing,
        tuning_run=best_tuned.run_id,
    ),
)

result.metrics.model_dump()
```

## 3. Analyse

Die Auswahl der Monatsdarstellung wird anhand der CV-Tabelle eingeordnet. Der gewählte Kandidat wird gegen die aktuelle RBF-SVM-Referenz verglichen; die Referenz bleibt nur bei einem Validierungsgewinn ersetzt.

```python
from awp2.evaluation import plot_confusion_matrices
from awp2.plots import plot_class_recall, plot_top_confusions, plot_tuning_candidates

REFERENCE_SVM_CV_BACC_COMBINED = 0.8473
REFERENCE_SVM_VALIDATION_BACC_COMBINED = 0.8664

comparison = pd.DataFrame(
    {
        "CV BAcc kombiniert": {
            "RBF-SVM Referenz": REFERENCE_SVM_CV_BACC_COMBINED,
            "Beste Month-Kodierung": best_tuned.best_score,
        },
        "Validierung BAcc kombiniert": {
            "RBF-SVM Referenz": REFERENCE_SVM_VALIDATION_BACC_COMBINED,
            "Beste Month-Kodierung": result.metrics.bacc_combined,
        },
    }
)
display(comparison)

display(cv_summary)

tuning_ax = plot_tuning_candidates(
    best_tuned.candidates,
    color_by="estimator__C",
    validation_score=result.metrics.bacc_combined,
)
tuning_ax.set_title(f"Kombinierte Labels: RBF-SVM mit {best_month_encoding} Month-Kodierung")
display(tuning_ax.figure)

for figure in (
    plot_confusion_matrices(result.y_val, result.y_pred),
    plot_class_recall(result.y_val, result.y_pred),
    plot_top_confusions(result.y_val, result.y_pred),
):
    display(figure)
```

## Fazit

Die numerische Darstellung von `Month` gewinnt die CV gegen One-Hot und zyklische Kodierung. Ihr einmaliges Validierungsergebnis liegt jedoch unter der bestehenden RBF-SVM-Referenz. Die Hypothese wird daher nicht gestützt: Die Referenz behält die numerische Kodierung mit `C=100.0` und `gamma=0.003`.
